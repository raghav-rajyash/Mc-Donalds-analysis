```python
fig, axes = plt.subplots(nrows=3,ncols=3)
fig.set_size_inches(20, 20)

a = sns.boxplot(data=McD[['Calories','Calories from Fat','Total Fat','Total Fat (% Daily Value)']] ,
                orient = "v" , ax=axes[0][0])

b = sns.boxplot(data=McD[['Saturated Fat','Saturated Fat (% Daily Value)','Trans Fat']] ,
                orient = "v" , ax=axes[0][1])

c = sns.boxplot(data=McD[['Cholesterol','Cholesterol (% Daily Value)']] ,
                orient = "v" , ax=axes[0][2])
d = sns.boxplot(data=McD[['Calories','Calories from Fat','Total Fat','Total Fat (% Daily Value)']] ,
                orient = "v" , ax=axes[1][0])
e = sns.boxplot(data=McD[['Sodium','Sodium (% Daily Value)']] ,
                orient = "v" , ax=axes[1][1])
f = sns.boxplot(data=McD[['Carbohydrates','Carbohydrates (% Daily Value)']] ,
                orient = "v" , ax=axes[1][2])
g = sns.boxplot(data=McD[['Dietary Fiber','Dietary Fiber (% Daily Value)']] ,
                orient = "v" , ax=axes[2][0])
h = sns.boxplot(data=McD[['Sugars','Protein']] ,
                orient = "v" , ax=axes[2][1])
i = sns.boxplot(data=McD[['Vitamin A (% Daily Value)','Vitamin C (% Daily Value)', 'Calcium (% Daily Value)', 'Iron (% Daily Value)']] ,
                orient = "v" , ax=axes[2][2])


plt.show()
```

```python
numeric_columns = McD.select_dtypes(include=['float64', 'int64']).columns
cor = McD[numeric_columns].corr()

plt.figure(figsize=(15, 15))
sns.heatmap(cor, annot=True, vmin=-1, vmax=1, center=0)
plt.show()
```
